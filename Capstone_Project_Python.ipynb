```python
import pandas as pd
df = pd.read_csv("Capstone Project Python - Sheet1.csv", header=1)
print(df)
```

```python
# Table 1
df1 = df.iloc[1:14, 0:4]

# Table 2
table2 = df.iloc[17:23, 0:5]

# Table 3
table3 = df.iloc[26:32, 0:2]


print(table1)
print(table2)
print(table3)
```

**#Task 2**
The cost column in the dataframe “Project” has some missing values. Your task is to
compute these missing values. Replace the missing values by running average. You
should use the “For” loop for this task.

```python
# Convert cost column to numeric
table1['Cost'] = pd.to_numeric(table1['Cost'], errors='coerce')

# Apply running average using FOR loop
for i in range(len(table1)):

    if pd.isna(table1.loc[i, 'Cost']):

        if i == 0:
            table1.loc[i, 'Cost'] = 0   # handle first row
        else:
            avg = table1.loc[:i-1, 'Cost'].mean()
            table1.loc[i, 'Cost'] = avg

print(table1)
```

**#Task3**
Split the name column in the Employee dataframe into two new columns “First Name”,
and “LastName” and remove the older “name” column.

```python
header2 = df.iloc[17,0:5]

# Data (rows after header)
table2 = df.iloc[18:23, 0:5]

# Assign header
table2.columns = header2

# Split 'name' column into two parts
table2[['First Name', 'LastName']] = table2['Name'].str.split(' ', expand=True)

# Drop the original 'name' column
table2.drop('Name', axis=1, inplace=True)

print(table2)
```

```python
header3 = df.iloc[26,0:2]

# Data (rows after header)
table3 = df.iloc[27:32, 0:2]

# Assign header
table3.columns = header3
print(table3)
```

**#Task 4**
Join all three dataframes in one single dataframe. Name it “Final”

```python
# Merge table1 and table2
temp = pd.merge(table1, table2, on='ID', how='outer')

# Merge result with table3
Final = pd.merge(temp, table3, on='ID', how='outer')

print(Final)
```

**#Task 5**
Add a new bonus column in the Final dataframe. Give a 5% bonus concerning project
cost only to employees who have finished the projects.

```python
# Create Bonus column (default = 0)
Final['Bonus'] = 0

# Apply 5% bonus only where project is finished
Final.loc[Final['Status'] == 'Finished', 'Bonus'] = Final['Cost'] * 0.05

print(Final)
```

**#Task 6**
Demote the designation level by 1, whose projects have status “fail”. Delete the
employees record whose designation level is above 4.


```python
# Demote where Status = Fail
Final.loc[Final['Status'] == 'Failed', 'Designation Level'] -= 1

Final['Designation Level'] = Final['Designation Level'].clip(lower=1)

# Remove employees with designation > 4
Final = Final[Final['Designation Level'] <= 4]

print(Final)
```

**#Task 7**
Add “Mr.” and “Mrs.” to the first name column and drop the gender column.

```python
# Add prefix based on Gender
Final.loc[Final['Gender'] == 'M', 'First Name'] = 'Mr. ' + Final['First Name']
Final.loc[Final['Gender'] == 'F', 'First Name'] = 'Mrs. ' + Final['First Name']

# Drop Gender column
Final = Final.drop('Gender', axis=1)

print(Final)
```

**#Task 8**
Promote designation level by 1 for the employees whose age is more than 29 years using
IF condition.

```python
# Final['Age'] > 29 → condition to filter employees
# loc[...] → selects only those rows and updates the column
Final.loc[Final['Age'] > 29, 'Designation Level'] += 1

print(Final)
```

**#Task 9**
Add the cost of all projects for each Employee and save it in new dataframe
“TotalProjCost” with three columns ID, First Name, and Total cost

```python
# Group the data by 'ID' and 'First Name'
# This ensures each employee is grouped together
TotalProjCost = Final.groupby(
    ['ID', 'First Name']
)['Cost'].sum().reset_index()# Sum all project costs and convert result back to DataFrame

# Rename the column 'Cost' to 'Total Cost' for better readability
TotalProjCost = TotalProjCost.rename(columns={'Cost': 'Total Cost'})

print(TotalProjCost)
```

**#Task 10**
Print all the employee details whose city name contains the letter “o” in it.


```python
# Using apply and lambda function

filtered_df = Final[Final['City'].apply(
    lambda x: 'o' in str(x).lower()   # convert to string and check 'o'
)]

print(filtered_df)
```
